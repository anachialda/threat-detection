## Random Forest pe CIC-IDS2017

```python
import time
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score, accuracy_score
from pathlib import Path

data_path = Path('data/Wednesday-WorkingHours.csv')

seeds = [1, 2, 3]          
test_size = 0.3           
Path('results').mkdir(exist_ok=True)
```

### 1. Citim datele

```python
df = pd.read_csv(data_path, low_memory=False)
df.columns = df.columns.str.strip()
print('Reihen, Spalten:', df.shape)

# spalte mit label
label = [c for c in df.columns if c.lower() == 'label'][0]
df[label] = df[label].astype(str).str.strip()
df[label].value_counts()
```

```python
attempted = df[label].str.contains('attempted', case=False)
print('Attempted Reihen geloscht:', attempted.sum())
df = df[~attempted].copy()
```

```python
#Spalten, die identifizieren

id_cols = ['id', 'Flow ID', 'Src IP', 'Source IP', 'Dst IP', 'Destination IP',
           'Src Port', 'Source Port', 'Timestamp', 'Attempted Category']
drop_cols = [c for c in df.columns if c in id_cols]
print('Spalten geloscht:', drop_cols)
df = df.drop(columns=drop_cols)

features = df.drop(columns=[label]).select_dtypes(include='number').columns.tolist()
df = df[features + [label]]
print('Nr. features:', len(features))
```

```python
#Reihen mit fehlenden Werten
df = df.replace([np.inf, -np.inf], np.nan)
print('Reihen mit fehlenden Werte:', df.isna().any(axis=1).sum())
df = df.dropna()
```

```python
# Duplikate
before = len(df)
df = df.drop_duplicates()                                  # identisch
df = df[~df.duplicated(subset=features, keep=False)]       # gleiche werte, verschiedene labels
print(f'Reihen bevor: {before}, nach: {len(df)}')

constant = [c for c in features if df[c].nunique() <= 1]
features = [c for c in features if c not in constant]
print('Konstante Spalten geloscht:', len(constant))
```

```python
#binare label
X = df[features]
y = (df[label].str.upper() != 'BENIGN').astype(int)
print(y.value_counts().rename({0: 'BENIGN (0)', 1: 'Atac (1)'}))
```

```python
def evalueaza(y_true, y_pred):
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred, labels=[0, 1]).ravel()
    return {
        'FPR': fp / (fp + tn),
        'Recall': recall_score(y_true, y_pred, zero_division=0),
        'Precision': precision_score(y_true, y_pred, zero_division=0),
        'F1': f1_score(y_true, y_pred, zero_division=0),
        'Accuracy': accuracy_score(y_true, y_pred),
        'TN': tn, 'FP': fp, 'FN': fn, 'TP': tp,
    }

rezultate = []
for seed in seeds:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, stratify=y, random_state=seed)

    rf = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=seed)
    t0 = time.perf_counter(); rf.fit(X_train, y_train); t_train = time.perf_counter() - t0
    t0 = time.perf_counter(); y_pred = rf.predict(X_test); t_pred = time.perf_counter() - t0

    r = evalueaza(y_test, y_pred)
    r.update({'seed': seed, 'train_s': t_train, 'ms_pe_flow': 1000 * t_pred / len(X_test)})
    rezultate.append(r)
    print(f"seed {seed}: FPR={r['FPR']:.4f}  Recall={r['Recall']:.4f}  F1={r['F1']:.4f}  "
          f"{r['ms_pe_flow']:.4f} ms/flow")

rez = pd.DataFrame(rezultate)
rez.to_csv('results/rf_rezultate.csv', index=False)
```

### medie ± abatere standard

```python
metrici = ['FPR', 'Recall', 'Precision', 'F1', 'Accuracy', 'ms_pe_flow']
rezumat = rez[metrici].agg(['mean', 'std']).T
rezumat
```

```python
# testen auf noch nicht gesehene angrifftypen
etichete = df[label]
tipuri = etichete[etichete.str.upper() != 'BENIGN'].value_counts()
tipuri = tipuri[tipuri >= 50].index.tolist()
benign_idx = etichete[etichete.str.upper() == 'BENIGN'].index
b_train, b_test = train_test_split(benign_idx, test_size=test_size, random_state=1)

loao = []
for tip in tipuri:
    vazut = etichete[(etichete.str.upper() != 'BENIGN') & (etichete != tip)].index
    nevazut = etichete[etichete == tip].index
    rf_l = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=1)

    t0 = time.perf_counter()
    rf_l.fit(X.loc[b_train.union(vazut)], y.loc[b_train.union(vazut)])
    t_train = time.perf_counter() - t0

    idx_test = b_test.union(nevazut)
    t0 = time.perf_counter()
    pred = pd.Series(rf_l.predict(X.loc[idx_test]), index=idx_test)
    t_pred = time.perf_counter() - t0

    loao.append({'atac_nevazut': tip, 'flow-uri': len(nevazut),
                 'Recall_nevazut': pred.loc[nevazut].mean(),
                 'FPR': pred.loc[b_test].sum()/len(b_test),
                 'F1': f1_score(y.loc[idx_test], pred),
                 'timp_antrenare_s': t_train,
                 'ms_pe_flow': 1000 * t_pred / len(idx_test)})

loao = pd.DataFrame(loao).round(4)
loao.to_csv('results/rf_atac_nevazut.csv', index=False)
print(loao)
```
